```python
import numpy as np

# ----- Define A -----
A = (1/10) * np.array([
    [ 5, -1, -2,  0],
    [-1,  5,  0, -2],
    [-2,  0,  5, -1],
    [ 0, -2, -1,  5]
], dtype=float)

# ----- Define b -----
b = np.array([1, 0, 0, 0], dtype=float)

# ----- Compute eigenvalues & eigenvectors -----
# np.linalg.eigh returns eigenvalues in ascending order
eigvals, eigvecs = np.linalg.eigh(A)

# ----- Factor out 0.2 and express eigenvalues as nice fractions -----
# We know 0.2 = 1/5, so eigenvalues should be k/5 for k = 1,2,3,4
scaled_eigvals = np.round(eigvals / 0.2).astype(int)  # k ≈ 1,2,3,4

print("Eigenvalues of A (as fractions k/5):")
for i, k in enumerate(scaled_eigvals, start=1):
    frac_str = f"{k}/5"
    print(f"λ{i} = {frac_str}  (numeric ≈ {eigvals[i-1]:.6f})")

# ----- Print eigenvector matrix -----
print("\nEigenvector matrix (columns are eigenvectors):")
print(eigvecs)

# ----- Sanity checks on eigenvectors -----
print("\nCheck: ||A v_i - λ_i v_i|| for each eigenpair:")
for i, (lam, v) in enumerate(zip(eigvals, eigvecs.T), start=1):
    residual = np.linalg.norm(A @ v - lam * v)
    print(f"i = {i}: residual ≈ {residual:.3e}")

# Check orthonormality of eigenvectors: V^T V should be ~ I
VtV = eigvecs.T @ eigvecs
print("\nV^T V (should be close to identity):")
print(VtV)
```

```python
import numpy as np

# This cell assumes:
#   - b is your 4D real vector (shape: (4,))
#   - eigvecs is the 4x4 eigenvector matrix from np.linalg.eigh(A)
#     (columns are the eigenvectors |u_i>)

# 1. Normalize b → |b>
norm_b = np.linalg.norm(b)
ket_b = b / norm_b

print("Normalized state |b> (in computational basis):")
print(ket_b)
print()

# 2. Expansion coefficients c_i = ⟨u_i|b⟩
#    Columns of eigvecs are |u_i>, so eigvecs.T.conj() @ |b> gives all c_i
c = eigvecs.T.conj() @ ket_b

print("|b> expanded in the eigenbasis {|u_i>} of A:")
print("    |b> = Σ_i c_i |u_i>")
print("Coefficients c_i (⟨u_i|b⟩):")
for i, ci in enumerate(c, start=1):
    # For clarity, separate real/imag if needed
    if abs(ci.imag) < 1e-10:
        # Purely real (or almost) → print as real
        print(f"  c_{i} = {ci.real:.6f}")
    else:
        # Complex → a + ib format
        print(f"  c_{i} = {ci.real:.6f} + {ci.imag:.6f} i")
print()

# 3. Check normalization in eigenbasis: Σ |c_i|^2 should be 1
prob_sum = np.sum(np.abs(c)**2)
print("Sum of probabilities Σ_i |c_i|^2 =", f"{prob_sum:.6f}")
print()

# 4. Reconstruct |b> from the eigenbasis to verify consistency
reconstructed_b = eigvecs @ c

print("Reconstructed |b> from Σ_i c_i |u_i>:")
print(reconstructed_b)
print()

diff_norm = np.linalg.norm(reconstructed_b - ket_b)
print("‖ |b> - reconstructed |b> ‖ =", f"{diff_norm:.3e}")
```

HHL circuit

```python
# ==============================
# Cell 2 – Registers, QFT/QPE, HHL circuit
# ==============================

from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit.circuit.library import UnitaryGate

# This cell assumes:
#   - A, eigvals, eigvecs, b are already defined in previous cells.
#   - eigvals, eigvecs come from np.linalg.eigh(A).

# --- 1. Registers ---

ancilla = QuantumRegister(1, "anc")
clock   = QuantumRegister(3, "clock")   # 3 phase/qubit register
b_reg   = QuantumRegister(2, "b")       # 2-qubit system (|b>)
creg    = ClassicalRegister(3, "c")     # c[0] for ancilla, c[1:3] for system

circuit = QuantumCircuit(ancilla, clock, b_reg, creg)

circuit.barrier(label="start")

# --- 2. Inverse QFT on 3 clock qubits (QFT†) ---

def qft_dagger(circ, q):
    """Inverse QFT on 3-qubit clock register q[0], q[1], q[2]."""
    # Order: q[0] (MSB), q[1], q[2] (LSB)

    # LSB branch
    circ.h(q[2])
    circ.cp(-np.pi / 2, q[1], q[2])
    circ.cp(-np.pi / 4, q[0], q[2])

    circ.h(q[1])
    circ.cp(-np.pi / 2, q[0], q[1])

    circ.h(q[0])

    # Reverse qubit order
    circ.swap(q[0], q[2])

# --- 3. QFT on 3 clock qubits (QFT) ---

def qft(circ, q):
    """QFT on 3-qubit clock register q[0], q[1], q[2]."""
    circ.swap(q[0], q[2])

    circ.h(q[0])
    circ.cp(np.pi / 2, q[0], q[1])
    circ.h(q[1])

    circ.cp(np.pi / 4, q[0], q[2])
    circ.cp(np.pi / 2, q[1], q[2])
    circ.h(q[2])

# --- 4. Controlled U^powers based on A, eigvals, eigvecs ---

t = 5 * np.pi / 4  # t = 10π/8 gives nice phases {1/8, 2/8, 3/8, 4/8}

def controlled_U_power(circ, clock_qubit, system_reg, power, sign=+1):
    """
    Apply controlled-U^power, where U = exp(i * sign * A * t),
    using the eigen-decomposition A = V diag(λ) V†.
    system_reg is the 2-qubit register for the 4D system.
    """
    phases = np.exp(1j * sign * eigvals * t * power)
    U = eigvecs @ np.diag(phases) @ eigvecs.conj().T

    U_gate = UnitaryGate(U, label=f"U^{power}" if sign > 0 else f"U^-{power}")
    circ.append(U_gate.control(1), [clock_qubit, system_reg[0], system_reg[1]])

def apply_controlled_U_powers(circ, clock_reg, system_reg, sign=+1):
    """
    Apply controlled U^(2^k) for k = 0,1,2 (3 clock qubits),
    where U = exp(i * sign * A * t).
    sign = +1: forward powers   (for QPE)
    sign = -1: adjoint powers   (for inverse QPE)
    """
    controlled_U_power(circ, clock_reg[0], system_reg, power=1, sign=sign)
    controlled_U_power(circ, clock_reg[1], system_reg, power=2, sign=sign)
    controlled_U_power(circ, clock_reg[2], system_reg, power=4, sign=sign)

# --- 5. QPE and inverse QPE ---

def apply_eigenvalue_controlled_ry(circ, clock_reg, anc_reg, eigvals, C=None):
    """
    Apply R_y(θ_i) to ancilla for each eigenvalue λ_i,
    conditioned on the corresponding 3-bit clock pattern.

    Assumes:
      - eigvals are [1/5, 2/5, 3/5, 4/5] (sorted ascending)
      - QPE maps them to clock states |001>, |010>, |011>, |100>
    """
    if C is None:
        C = np.min(eigvals)  # here C = 1/5

    # Map each λ_i to its clock basis pattern from QPE
    patterns = ["001", "010", "011", "100"]

    for lam, pattern in zip(eigvals, patterns):
        ratio = C / lam
        if ratio > 1.0:
            raise ValueError(f"C/λ = {ratio} > 1 for λ={lam}, choose smaller C.")

        theta = 2 * np.arcsin(ratio)

        # We want a multi-controlled R_y on ancilla, triggered when clock == pattern.
        # To do this with mcry, we:
        #   - X-flip any qubit whose pattern bit is '0' so desired state → |111>
        #   - apply mcry(theta, [clock_qubits], ancilla)
        #   - uncompute the X flips

        controls = []
        flipped_indices = []

        for i, bit in enumerate(pattern):
            if bit == "0":
                circ.x(clock_reg[i])
                flipped_indices.append(i)
            controls.append(clock_reg[i])

        # Multi-controlled R_y(θ) on ancilla, with all 3 clock qubits as controls
        circ.mcry(theta, controls, anc_reg[0], None)

        # Uncompute the X gates
        for i in flipped_indices:
            circ.x(clock_reg[i])


# --- 6. HHL macro ---

def hhl(circ, anc_reg, clock_reg, system_reg, classical_reg):
    """
    HHL macro:
      1) QPE
      2) Eigenvalue-dependent R_y rotations on ancilla (via clock patterns)
      3) Measure ancilla (post-select on 1 in analysis)
      4) Inverse QPE
    """
    # 1) QPE: encode eigenvalue phases into clock
    qpe(circ, clock_reg, system_reg)

    circ.barrier(label="controlled R_y (eigenvalue dependent)")

    # 2) Apply R_y(θ_i) for each eigenvalue λ_i, using full 3-bit clock patterns
    apply_eigenvalue_controlled_ry(circ, clock_reg, anc_reg, eigvals)

    circ.barrier(label="measure ancilla")

    # 3) Measure ancilla right after rotations (as in reference)
    circ.measure(anc_reg[0], classical_reg[0])

    circ.barrier(label="inverse QPE")

    # 4) Inverse QPE to uncompute the clock
    inv_qpe(circ, clock_reg, system_reg)

    """
    HHL macro:
      1) QPE
      2) Controlled R_y from clock to ancilla
      3) Measure ancilla (post-select on 1 in analysis)
      4) Inverse QPE
    """
    # 1) QPE
    qpe(circ, clock_reg, system_reg)

    circ.barrier(label="controlled R_y")

    # 2) Controlled R_y rotations (heuristic angles; can be tuned using λ_i)
    theta0 = np.pi / 2    # control: clock[0]
    theta1 = np.pi / 3    # control: clock[1]
    theta2 = np.pi / 4    # control: clock[2]

    circ.cry(theta0, clock_reg[0], anc_reg[0])
    circ.cry(theta1, clock_reg[1], anc_reg[0])
    circ.cry(theta2, clock_reg[2], anc_reg[0])

    circ.barrier(label="measure ancilla")

    # 3) Measure ancilla *right after* R_y
    circ.measure(anc_reg[0], classical_reg[0])

    circ.barrier(label="inverse QPE")

    # 4) Inverse QPE (acts only on clock + system)
    inv_qpe(circ, clock_reg, system_reg)

# --- 7. State preparation for |b> ---

# For b = [1, 0, 0, 0], |b> = |00> on (b_0, b_1), so no gates needed.
circuit.barrier(label="prepare |b>")

# --- 8. Hadamards on clock (QPE initial step) ---

circuit.barrier(label="H on clock")
circuit.h(clock)

# --- 9. Run HHL block ---

hhl(circuit, ancilla, clock, b_reg, creg)

# --- 10. Optional: extra H on clock (to mirror reference style) ---

circuit.barrier(label="H on clock (after HHL)")
circuit.h(clock)

# --- 11. Measure system qubits |b> ---

circuit.barrier(label="measure system")
circuit.measure(b_reg[0], creg[1])
circuit.measure(b_reg[1], creg[2])

# --- 12. Draw circuit with Qiskit visualization ---

# Text ASCII circuit:
# circuit.draw(output="text")

# Pretty Matplotlib circuit:
circuit.draw(output="mpl")


```

```python
from qiskit_aer import AerSimulator
from qiskit import transpile
import numpy as np

sim = AerSimulator()

# Transpile circuit for simulator backend
tcirc = transpile(circuit, sim)

# Run simulation
result = sim.run(tcirc, shots=20000).result()
counts = result.get_counts()

print("Raw measurement counts:")
print(counts)

# --------------------------------------------------------
# POST-SELECTION: Keep only outcomes with ancilla = 1
# --------------------------------------------------------

post_counts = {}
total_post = 0

for bitstring, count in counts.items():
    # bitstring format: c2 c1 c0 (system + ancilla)
    # creg[0] = ancilla, creg[1], creg[2] = b_reg
    ancilla_bit = bitstring[-1]      # or bitstring[0] depending on order
    system_bits = bitstring[-3:]     # last 3 bits: ancilla + b1 + b0

    if ancilla_bit == '1':
        post_counts[bitstring] = count
        total_post += count

print("\nPost-selected (ancilla=1) counts:")
print(post_counts)
print(f"Total successful shots: {total_post}")

# --------------------------------------------------------
# Extract |x⟩ approximately (normalized amplitudes)
# --------------------------------------------------------

amps = np.zeros(4)

for bitstring, count in post_counts.items():
    # Extract system bits (b1 b0)
    sys = bitstring[-3:-1]  # last 2 bits before ancilla
    idx = int(sys, 2)
    amps[idx] += count

amps = amps / np.sum(amps)
amps = np.sqrt(amps)

print("\nEstimated |x⟩ from HHL (normalized):")
for i in range(4):
    print(f"  |{i:02b}> amplitude ≈ {amps[i]:.6f}")

# Compare to classical |x⟩
print("\nClassical |x⟩ (from earlier):")
print("  |00> =", ket_x[0])
print("  |01> =", ket_x[1])
print("  |10> =", ket_x[2])
print("  |11> =", ket_x[3])
```
