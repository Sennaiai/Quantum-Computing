# QFT Order Finding and Shor Factor Extraction

Controlled modular powers prepare $2^{-t/2}\sum_x|x\rangle|a^x\bmod N\rangle$; the inverse QFT converts periodic coherence into measurable phase peaks. Full states are compared with a separately constructed modular-power/FFT reference. Reconstruction proposes denominators from measured phases, verifies them modulo N, and retains unsuccessful factor-extraction cases.


```python
from pathlib import Path
import sys, json
ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p/'qft_core.py').is_file())
sys.path.insert(0, str(ROOT))
from qft_core import qft
from experiments import run_shor_experiments, plot_shor
print(qft(3, inverse=True).draw('text'))
report = run_shor_experiments()
```

## Executed outcomes

N=15 uses gate-built reversible QFT arithmetic, including clean support/overflow flags. N=21 uses a full compiled modular-permutation comparison, not a scalable arithmetic oracle. Each controlled power uses the stated base. No classical order-search fallback is used.


```python
for name in ('gate_arithmetic_N15','compiled_N15','compiled_N21_base2','compiled_N21_base5_negative_control'):
    row = report[name]
    print(name, {'order': row['recovery']['order'], 'factorization': row['factorization'],
        'state_error': row['state_error_l2_vs_exact_fft'],
        'workspace_leakage': row['workspace_leakage'], 'qubits': row['total_qubits']})
print('Gate/compiled phase-distribution agreement:', report['oracle_route_phase_TV_N15'])
print('Gate-built N=15 resources:', report['gate_arithmetic_N15_resources'])
```

## Inverse-QFT accuracy

The approximate inverse QFT is assessed against the exact FFT reference, not its own inverse. The figure compares ideal phase probabilities, seeded readout and an intentionally truncated transform. These are bounded ideal-simulator results; the [README](README.md) states oracle and resource limits.


```python
from IPython.display import SVG, display
approx = report['approximate_IQFT_N21']
print('Approximate inverse-QFT phase TV error:', approx['phase_tv_vs_exact_fft'])
print('Exact inverse-QFT resources:', report['exact_IQFT_resources'])
print('Approximate inverse-QFT resources:', report['approximate_IQFT_resources'])
figure = ROOT/'artifacts/notebook/order_finding.svg'
plot_shor(report, figure)
display(SVG(filename=str(figure)))
```
